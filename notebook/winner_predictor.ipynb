- <b>Total score in each innings in every matches.

```python
total_score_df = deliveries.groupby(['match_id','innings']).sum()['total_runs'].reset_index()
```

```python
print(total_score_df)
```

```python
total_score_df = total_score_df[total_score_df['innings']==1]
```

```python
total_score_df
```

```python
match_df = matches.merge(total_score_df[['match_id','total_runs']], left_on='match_id',right_on='match_id')
```

```python
match_df.head()
```

```python
match_df['team1'].unique()
```

```python
teams = [
    'Sunrisers Hyderabad',
    'Mumbai Indians',
    'Royal Challengers Bangalore',
    'Kolkata Knight Riders',
    'Punjab Kings',
    'Chennai Super Kings',
    'Rajasthan Royals',
    'Delhi Capitals',
    'Gujarat Titans',
    'Lucknow Super Giants'
]
```

```python
match_df['team1'] = match_df['team1'].str.replace('Delhi Daredevils','Delhi Capitals')
match_df['team2'] = match_df['team2'].str.replace('Delhi Daredevils','Delhi Capitals')

match_df['team1'] = match_df['team1'].str.replace('Deccan Chargers','Sunrisers Hyderabad')
match_df['team2'] = match_df['team2'].str.replace('Deccan Chargers','Sunrisers Hyderabad')
```

```python
match_df = match_df[match_df['team1'].isin(teams)]
match_df = match_df[match_df['team2'].isin(teams)]
```

```python
match_df.shape
```

```python
match_df['team1'].unique()
```

```python
match_df = match_df[['match_id','city','match_winner','total_runs']]
```

```python
delivery_df = match_df.merge(deliveries, on='match_id')
```

```python
delivery_df = delivery_df.merge(
    matches[['match_id', 'venue']],
    left_on='match_id',
    right_on='match_id',
    how='left'
)
```

```python
delivery_df
```

```python
delivery_df = delivery_df[delivery_df['innings']==2]
```

```python
delivery_df.info()
```

- <b> Runs left

```python
# delivery_df.groupby('match_id').cumsum()['total_runs_y']
delivery_df['current_score'] = delivery_df.groupby('match_id')['total_runs_y'].cumsum()
```

```python
delivery_df
```

```python
delivery_df['runs_left'] = delivery_df['total_runs_x'] - delivery_df['current_score'] + 1
```

```python
delivery_df
```

- <b> Balls Left

```python
delivery_df.columns
```

```python
delivery_df['over_number'].unique()
```

```python
delivery_df['balls_bowled'] = delivery_df['over_number'] * 6 + delivery_df['ball_number']
delivery_df['balls_left'] = 120 - delivery_df['balls_bowled']
```

```python
delivery_df
```

- <b> Wickets left

```python
delivery_df['wickets_fallen'] = delivery_df.groupby('match_id')['player_out'].transform(lambda x: x.notnull().cumsum())

delivery_df['wickets_left'] = 10 - delivery_df['wickets_fallen']
```

```python
delivery_df
```

- <b> Current Run Rate.

```python
# delivery_df['balls_bowled'] = (delivery_df['over_number'] * 6) + delivery_df['ball_number']
# delivery_df['crr'] = (delivery_df['current_score'] * 6) / delivery_df['balls_bowled']

# over number is zero at starting of inning, so handling it can be done by.

delivery_df['balls_bowled'] = (delivery_df['over_number'] * 6) + delivery_df['ball_number']

delivery_df['crr'] = (delivery_df['current_score'] * 6) / delivery_df['balls_bowled'].replace(0, 1)
```

```python
delivery_df
```

- <b> Required Run Rate

```python
delivery_df['rrr'] = (delivery_df['runs_left'] * 6) / delivery_df['balls_left'].replace(0, 1)
```

```python
delivery_df
```

- <b> Result

```python
def result(row):
    return 1 if row['team_batting']== row['match_winner'] else 0
```

```python
delivery_df['results'] = delivery_df.apply(result, axis=1)
```

```python
delivery_df
```

- <b>Extract useful columns from delivery dataset

```python
final_df = delivery_df[['team_batting','team_bowling','venue','runs_left','balls_left','wickets_left','total_runs_x','crr','rrr','results']]
```

```python
final_df.dropna(inplace=True)

final_df.shape
```

```python
final_df = final_df.sample(final_df.shape[0])
```

```python
final_df.sample()
```

- <b> Model Development

```python
x = final_df.iloc[:,:-1]
y = final_df.iloc[:,-1]
```

```python
import pandas as pd
import numpy as np
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

categorical_cols = ['team_batting', 'team_bowling', 'venue']

preprocessor = ColumnTransformer(transformers=[('cat', OneHotEncoder(sparse_output=False, drop='first', handle_unknown='ignore'), categorical_cols)],remainder='passthrough')
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('model', LogisticRegression(max_iter=1000, solver='liblinear'))
])
```

```python
pipeline.fit(x_train,y_train)
```

```python
y_pred = pipeline.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report

accuracy_score(y_test,y_pred)
```
